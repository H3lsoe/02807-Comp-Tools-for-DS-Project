```python
import os
import pandas as pd
import folium
import seaborn as sns
import matplotlib.pyplot as plt

cwd = os.getcwd()
if os.path.basename(cwd) == "Notebooks":
    project_root = os.path.dirname(cwd)
else:
    project_root = cwd
path = os.path.join(project_root, "data", "Airbnb_DK_Processed_Data.csv")
df = pd.read_csv(path)
```

```python
import pandas as pd
import folium
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. Create subsets
# ============================================================

# Subset 1: availability > 70 days
avail_over_70 = df[df['availability_365'] > 70].copy()

# Subset 2: hosts with more than one listing
host_counts = df['host_id'].value_counts()
multi_hosts = host_counts[host_counts > 1].index
multi_listing = df[df['host_id'].isin(multi_hosts)].copy()

# ============================================================
# 2. Build Folium map centered on mean location
# ============================================================
fig = folium.Figure(width=800, height=600)
m = folium.Map(
    location=[df['latitude'].mean(), df['longitude'].mean()],
    zoom_start=12#,
    # width=800,   # pixels or '50%'
    # height=600   # pixels
)

# ============================================================
# 3. Define colors for the two groups
# ============================================================

colors = {
    "avail_over_70": "rgba(0, 0, 255, 0.6)",     # blue
    "multi_listing": "rgba(255, 0, 0, 0.6)"      # red
}

# ============================================================
# 4. Plot availability > 70 days (blue markers)
# ============================================================

for _, row in avail_over_70.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=3,
        color=colors["avail_over_70"],
        fill=True,
        fill_color=colors["avail_over_70"],
        fill_opacity=0.3,
        opacity=0.5,
        popup="Availability > 70 days"
    ).add_to(m)

# ============================================================
# 5. Plot multi-listing hosts (red markers)
# ============================================================

for _, row in multi_listing.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=3,
        color=colors["multi_listing"],
        fill=True,
        fill_color=colors["multi_listing"],
        fill_opacity=0.3,
        opacity=0.5,
        popup="Host with > 1 listing"
    ).add_to(m)

# ============================================================
# 6. Display the map
# ============================================================
fig.add_child(m)
fig
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# --- 1. Define your subset conditions ---
subset_condition = (
    (df['price'] > 1500) &
    (df['estimated_occupancy_l365d'] > 2) 
)

df_subset = df[subset_condition].copy()

# --- 2. Define the features to use for clustering ---
features = [
    'price',
    'estimated_occupancy_l365d',
    # 'minimum_nights',
    # 'number_of_listings_by_host',  # if computed before
    # add more numeric features as needed
]

# Compute number_of_listings_by_host if not present
if 'number_of_listings_by_host' not in df_subset.columns:
    host_counts = df_subset['host_id'].value_counts()
    df_subset['number_of_listings_by_host'] = df_subset['host_id'].map(host_counts)

# Drop rows with missing feature values
df_subset = df_subset.dropna(subset=features)

# --- 3. Standardize features ---
scaler = StandardScaler()
X = scaler.fit_transform(df_subset[features])

# --- 4. Run K-Means ---
n_clusters = 2  # Adjust as needed
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)

# Store cluster labels
df_subset['cluster'] = labels
```

```python
import folium
import matplotlib.pyplot as plt

# Base map
fig_cluster = folium.Figure(width=800, height=600)
m = folium.Map(
    location=[df_subset['latitude'].mean(), df_subset['longitude'].mean()],
    zoom_start=12,
    # width=600,
    # height=400
)

# Generate unique colors for each cluster
clusters = sorted(df_subset['cluster'].unique())
cmap = plt.cm.get_cmap('tab10', len(clusters))
colors = {
    0: "#FF0000",  # red
    1: "#0000FF",  # blue
    # 2: "#00FF00",  # green
    # 3: "#FFA500",  # orange
    # 4: "#800080",  # purple
}
cluster_labels = {
    0: "Commercial",
    1: "Residential"
}

# Plot points colored by cluster
for _, row in df_subset.iterrows():
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=4,
        color=colors[row['cluster']],
        fill=True,
        fill_color=colors[row['cluster']],
        fill_opacity=0.3,
        opacity=0.3,
        popup=f"Cluster: {row['cluster']}"
    ).add_to(m)

# Display map
fig_cluster.add_child(m)
fig_cluster
```
